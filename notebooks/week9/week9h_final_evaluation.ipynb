## H2 — Explicit one-shot confirmation

Enable **only after H0 passes**. This executes pre-frozen producers and scores complete, aligned TEST exports using the common evaluator. All mandatory comparators are required. A sealed confirmation cannot run again; H3 consumes its saved predictions.

The v2 producers prepare TEST retrieval/SFR and the canonical first pass, then run the frozen static, W9 and recovered W8 systems sequentially. This cell automatically adds --resume when an existing H2 claim is present. Completed producers are checksum-checked and skipped; partial producers reuse their saved query shards and generate only missing queries. Progress counters may start at 1 while checking cached queries; this does not mean answers are regenerated. If H2 is already sealed, this cell skips generation and leaves H3 available.

```python
if RUN_FINAL_CONFIRMATION:
    if (RESULT_ROOT/'h2_seal.json').exists():
        print('Confirmation already sealed; skipping H2 generation. Run H3 for the report.')
    else:
        resume = (RESULT_ROOT/'h2_claim.json').exists()
        flags = ['--run-final-confirmation']
        if resume:
            flags.append('--resume')
            print('Resuming the existing frozen run; completed answers will not be regenerated.')
            completed = sorted(path.stem for path in (RESULT_ROOT/'receipts').glob('*.json'))
            print('Completed producer receipts:', ', '.join(completed) or 'none yet')
            for folder in sorted((RESULT_ROOT/'producer_cache').glob('*')):
                if folder.is_dir():
                    saved = sum(1 for path in folder.glob('*.json') if len(path.stem) == 64)
                    print(f'  {folder.name}: {saved} saved query results')
            print('Freeze/source checks run first; cached progress counters can restart at 1 without generation.')
        else:
            print('Starting the first confirmation run; no prior H2 claim exists.')
        stage('h2', *flags)
else:
    print('Final confirmation disabled. No TEST files or models opened.')
if RUN_ROBUSTNESS:
    raise ValueError('No new model panel may be selected; robustness is not enabled by this protocol.')
```
