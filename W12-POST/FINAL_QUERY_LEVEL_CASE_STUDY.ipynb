## 5. Browse exact case dossiers

Each dossier includes the query, diagnostic gold annotations, frozen retrieved pool, compact four-column method comparison, exact reader contexts, selected source units, retained/discarded sentences, saved fidelity/IG/P_help/stop traces, and candidate mechanisms. Only actual saved computations are shown.

Use the selector below if widgets are available. Otherwise change `CASE_ID` and rerun this display cell. This reads an existing report; it does not rerun analysis.


```python
CASE_ID = 'C01'
report_path = OUT/'CASE_STUDIES.md'
if report_path.exists():
    import re
    content = report_path.read_text()
    parts = re.split(r'(?m)^(?=## C[0-9]+ ·)', content)
    dossiers = {re.match(r'## (C[0-9]+)', part).group(1): part
                for part in parts if re.match(r'## C[0-9]+', part)}
    def show_case(case_id):
        display(Markdown(dossiers.get(case_id, 'Unknown case ID; see CASE_SELECTION.csv.')))
    try:
        import ipywidgets as widgets
        selector = widgets.Dropdown(options=list(dossiers), value=CASE_ID if CASE_ID in dossiers else next(iter(dossiers)), description='Case:')
        viewer = widgets.Output()
        def refresh_case(change=None):
            with viewer:
                viewer.clear_output(wait=True)
                show_case(selector.value)
        selector.observe(refresh_case, names='value')
        display(selector, viewer)
        refresh_case()
    except ImportError:
        show_case(CASE_ID)
else:
    print('Run the saved-result analysis first.')
```
